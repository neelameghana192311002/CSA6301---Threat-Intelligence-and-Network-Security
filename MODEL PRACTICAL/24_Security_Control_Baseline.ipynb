```python
controls = {
    "password": (8, 10),
    "backup": (1, 1),
    "firewall": (1, 1),
    "patch": (70, 90)
}

for control, values in controls.items():
    current, required = values

    if current < required:
        print(control, "FAIL")
    else:
        print(control, "PASS")
```
