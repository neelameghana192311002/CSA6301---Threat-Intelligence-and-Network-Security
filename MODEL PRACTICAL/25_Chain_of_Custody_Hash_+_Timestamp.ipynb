```python
records = [
    {"hash":"A", "time":1},
    {"hash":"B", "time":2},
    {"hash":"C", "time":3}
]

ok = True

for i in range(1, len(records)):
    if records[i]["time"] < records[i-1]["time"]:
        ok = False

if ok:
    print("Chain is valid")
else:
    print("Chain is broken")
```
