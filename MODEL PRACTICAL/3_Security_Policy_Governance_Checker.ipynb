```python
policy = input("Enter security policy text: ").lower()

requirements = {
    "scope": "scope",
    "roles and responsibilities": "roles",
    "enforcement": "enforcement",
    "review cycle": "review",
    "approval authority": "approval"
}

for item, keyword in requirements.items():

    if keyword in policy:
        print(item, ": Present")
    else:
        print(item, ": Missing")
```
