```python
logs = [
    "192.168.1.10 LOGIN SUCCESS",
    "192.168.1.20 LOGIN FAILED",
    "192.168.1.10 LOGIN SUCCESS",
    "192.168.1.30 LOGIN FAILED"
]

events = []

# Logstash-style parsing
for log in logs:

    parts = log.split()

    event = {
        "ip": parts[0],
        "action": parts[1],
        "status": parts[2]
    }

    events.append(event)

# Elasticsearch-style indexing
index = events

# Kibana-style aggregation
success = 0
failed = 0

for event in index:

    if event["status"] == "SUCCESS":
        success += 1
    else:
        failed += 1

print("Structured Events:")
for event in index:
    print(event)

print("\nKibana-style Summary")
print("Successful logins:", success)
print("Failed logins:", failed)
```
