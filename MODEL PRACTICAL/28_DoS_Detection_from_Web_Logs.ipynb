```python
logs = [
    "10.0.0.1",
    "10.0.0.1",
    "10.0.0.1",
    "10.0.0.1",
    "10.0.0.1",
    "10.0.0.2"
]

count = {}

for ip in logs:
    count[ip] = count.get(ip, 0) + 1

for ip, requests in count.items():
    if requests > 3:
        print("Possible DoS attack from", ip)
```
