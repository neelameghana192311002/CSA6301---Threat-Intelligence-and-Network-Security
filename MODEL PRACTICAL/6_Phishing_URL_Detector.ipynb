```python
url = input("Enter URL: ").lower()

suspicious_words = ["login", "verify", "update", "free", "password"]

score = 0

# Check suspicious keywords
for word in suspicious_words:
    if word in url:
        score += 1

# Check IP address
parts = url.replace("http://", "").replace("https://", "").split("/")

domain = parts[0]

if domain.replace(".", "").isdigit():
    score += 2

# Check excessive subdomains
if domain.count(".") > 3:
    score += 1

print("Risk Score:", score)

if score >= 2:
    print("Result: Possible Phishing URL")
else:
    print("Result: URL appears normal")
```
