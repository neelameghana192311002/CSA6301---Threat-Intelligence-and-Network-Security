```python
control = input("Enter security control: ").lower()

if "access" in control or "password" in control:
    print("Domain: Access Control")

elif "cryptography" in control or "encryption" in control:
    print("Domain: Cryptography")

elif "physical" in control or "security camera" in control:
    print("Domain: Physical Security")

elif "supplier" in control or "vendor" in control:
    print("Domain: Supplier Relationships")

elif "incident" in control:
    print("Domain: Incident Management")

else:
    print("Domain: Other Annex A Control")
```
