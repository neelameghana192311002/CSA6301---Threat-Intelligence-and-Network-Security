```python
connections = []

# Outbound connection
source = input("Enter source: ")
destination = input("Enter destination: ")

connections.append((source, destination))

print("Outbound connection allowed")

# Return traffic
reply = input("Is this return traffic? (yes/no): ")

if reply == "yes" and (source, destination) in connections:
    print("Return traffic allowed")
else:
    print("Packet blocked")
```
