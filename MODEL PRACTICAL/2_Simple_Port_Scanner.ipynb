```python
import socket

host = input("Enter target host: ")

for port in range(1, 101):
    s = socket.socket()
    s.settimeout(0.5)

    result = s.connect_ex((host, port))

    if result == 0:
        print("Port", port, "is OPEN")

    s.close()
```
