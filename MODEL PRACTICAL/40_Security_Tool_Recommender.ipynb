```python
task = input("Enter security task: ").lower()

if "log" in task or "siem" in task:
    print("Recommended tool: Splunk")
elif "vulnerability" in task:
    print("Recommended tool: Nessus")
elif "packet" in task or "network" in task:
    print("Recommended tool: Wireshark")
elif "malware" in task:
    print("Recommended tool: Malwarebytes")
else:
    print("No recommendation")
```
