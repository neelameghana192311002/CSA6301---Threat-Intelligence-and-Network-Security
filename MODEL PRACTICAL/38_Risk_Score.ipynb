```python
likelihood = int(input("Enter likelihood (1-5): "))
impact = int(input("Enter impact (1-5): "))

risk = likelihood * impact

print("Risk score:", risk)

if risk >= 20:
    print("Critical")
elif risk >= 12:
    print("High")
elif risk >= 6:
    print("Medium")
else:
    print("Low")
```
