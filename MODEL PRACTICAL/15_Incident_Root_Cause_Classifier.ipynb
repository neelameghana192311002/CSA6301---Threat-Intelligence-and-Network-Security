```python
case = input("Enter incident description: ").lower()

if "phishing" in case or "email link" in case:
    print("Root Cause: Phishing")

elif "unpatched" in case or "vulnerability" in case:
    print("Root Cause: Unpatched Vulnerability")

elif "configuration" in case or "misconfigured" in case:
    print("Root Cause: Misconfiguration")

elif "employee" in case or "insider" in case:
    print("Root Cause: Insider Threat")

elif "vendor" in case or "third party" in case:
    print("Root Cause: Supply-Chain Compromise")

else:
    print("Root Cause: Unknown")
```
