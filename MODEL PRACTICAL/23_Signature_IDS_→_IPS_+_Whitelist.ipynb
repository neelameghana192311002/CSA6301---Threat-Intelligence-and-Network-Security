```python
source = input("Enter source IP: ")
traffic = input("Enter traffic: ").lower()

whitelist = ["192.168.1.10"]

if source in whitelist:
    print("Trusted source - ALLOW")
elif "attack" in traffic or "malware" in traffic:
    print("Threat detected - BLOCK")
else:
    print("ALLOW")
```
