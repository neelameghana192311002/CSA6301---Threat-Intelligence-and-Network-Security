```python
activity = input("Enter intelligence activity: ").lower()

if "requirement" in activity or "goal" in activity:
    print("Stage: Direction")

elif "collect" in activity or "gather" in activity:
    print("Stage: Collection")

elif "clean" in activity or "filter" in activity:
    print("Stage: Processing")

elif "analyze" in activity or "investigate" in activity:
    print("Stage: Analysis")

elif "report" in activity or "share" in activity:
    print("Stage: Dissemination")

elif "feedback" in activity or "review" in activity:
    print("Stage: Feedback")

else:
    print("Stage: Unknown")
```
