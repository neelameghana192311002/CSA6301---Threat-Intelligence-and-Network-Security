```python
attack = True

layers = ["Firewall", "IDS", "Antivirus", "Access Control"]

for layer in layers:

    answer = input(layer + " blocked attack? (yes/no): ")

    if answer == "yes":
        attack = False
        print(layer, "blocked the attack")
        break

if attack:
    print("Attack passed all layers")
else:
    print("System protected")
```
