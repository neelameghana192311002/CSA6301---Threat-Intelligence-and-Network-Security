```python
events = [
    {"user":"Alice", "status":"failed"},
    {"user":"Bob", "status":"success"},
    {"user":"Alice", "status":"failed"},
    {"user":"John", "status":"success"},
    {"user":"Alice", "status":"failed"}
]

keyword = input("Enter status to search: ")

filtered = [e for e in events if e["status"] == keyword]

print("Filtered events:", filtered)

count = {}
for e in events:
    user = e["user"]
    count[user] = count.get(user, 0) + 1

print("Count by user:", count)
```
