```python
report = input("Enter threat intelligence report: ").lower()
audience = input("Enter intended audience: ").lower()

if ("ceo" in audience or "management" in audience or
    "business" in report or "long term" in report):

    print("Classification: Strategic")

elif ("security team" in audience or "analyst" in audience or
      "attacker" in report or "tactic" in report):

    print("Classification: Tactical")

elif ("soc" in audience or "incident responder" in audience or
      "ip address" in report or "malware" in report):

    print("Classification: Operational")

else:
    print("Classification: Operational")
```
