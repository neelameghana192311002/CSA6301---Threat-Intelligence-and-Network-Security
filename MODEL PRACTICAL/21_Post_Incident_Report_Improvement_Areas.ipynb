```python
report = input("Enter post-incident report: ").lower()

areas = {
    "patch management": ["patch", "update"],
    "security awareness": ["training", "awareness"],
    "monitoring": ["monitoring", "logs"],
    "backup strategy": ["backup", "restore"],
    "access control": ["access", "permission"]
}

for area, words in areas.items():
    for word in words:
        if word in report:
            print("Found:", area)
            break
```
