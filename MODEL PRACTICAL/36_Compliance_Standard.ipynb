```python
text = input("Enter system/data description: ").lower()

if "credit card" in text or "payment" in text:
    print("PCI-DSS")
elif "patient" in text or "health" in text:
    print("HIPAA")
elif "personal data" in text or "privacy" in text:
    print("GDPR")
elif "financial report" in text or "accounting" in text:
    print("SOX")
else:
    print("Unknown")
```
