```python
logs = [
    "LOGIN user1 failed",
    "LOGIN user1 failed",
    "LOGIN user1 success"
]

# Collection
collected = logs

# Normalization
events = []

for log in collected:
    parts = log.split()

    events.append({
        "event": parts[0],
        "user": parts[1],
        "status": parts[2]
    })

# Correlation
failed = 0

for event in events:
    if event["status"] == "failed":
        failed += 1

# Alert
if failed >= 2:
    print("ALERT: Multiple failed logins detected")
else:
    print("No alert")
```
