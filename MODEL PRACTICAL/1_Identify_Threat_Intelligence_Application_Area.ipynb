```python
scenario = input("Enter business scenario: ").lower()

if "bank" in scenario or "fraud" in scenario or "payment" in scenario:
    print("Application Area: Banking Fraud Prevention")

elif "hospital" in scenario or "healthcare" in scenario or "patient" in scenario:
    print("Application Area: Healthcare Data Protection")

elif "shopping" in scenario or "ecommerce" in scenario or "online store" in scenario:
    print("Application Area: E-Commerce Fraud Prevention")

elif "government" in scenario:
    print("Application Area: Government Cybersecurity")

else:
    print("Application Area: General Cyber Threat Intelligence")
```
