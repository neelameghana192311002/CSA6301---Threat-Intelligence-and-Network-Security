```python
logs = [
    ("alice", "failed"),
    ("alice", "failed"),
    ("alice", "failed"),
    ("alice", "success")
]

user = input("Enter username: ")

failed = 0

for name, status in logs:
    if name == user:
        if status == "failed":
            failed += 1
        elif status == "success" and failed >= 3:
            print("Brute-force attack detected!")
            break
else:
    print("No brute-force attack")
```
