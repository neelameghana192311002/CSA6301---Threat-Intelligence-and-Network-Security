```python
rules = [
    ("allow", "HTTP"),
    ("deny", "HTTP"),
    ("allow", "SSH")
]

service = input("Enter service: ")

action = "deny"

for rule, name in rules:
    if name.lower() == service.lower():
        action = rule
        break

print("Action:", action.upper())
```
