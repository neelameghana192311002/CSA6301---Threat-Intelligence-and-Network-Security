```python
network = {
    "Low_Trust": ["Internet"],
    "User_Network": ["Low_Trust", "Server_Network"],
    "Server_Network": ["User_Network"],
    "Sensitive_Network": []
}

source = input("Enter source network: ")
destination = input("Enter destination network: ")

if destination in network[source]:
    print("Access Allowed")
else:
    print("Access Blocked")
```
